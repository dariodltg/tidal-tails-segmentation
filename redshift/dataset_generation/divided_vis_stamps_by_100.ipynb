```python
import numpy as np
import os
import pdb
from astropy.io import fits
from astropy import wcs
import subprocess
```

```python
folder_input_original_stamps = "stamps_capped_negative"
folder_output_stamps_divided_by_100 = "stamps_vis_divided_by_100"
```

```python
os.makedirs(folder_input_original_stamps, exist_ok=True)
os.makedirs(folder_output_stamps_divided_by_100, exist_ok=True)
```

```python
for file_name in os.listdir(folder_input_original_stamps):
    if file_name.endswith("_I_no_negatives.fits"):
        try:
            # Abrir el archivo FITS
            file_path = os.path.join(folder_input_original_stamps, file_name)
            hdu = fits.open(file_path)
            data = hdu[0].data
            hdr = hdu[0].header

            # Dividir los valores por 100
            data_divided_by_100 = data / 100.0

            # Definir la ruta del archivo de salida
            output_path = os.path.join(folder_output_stamps_divided_by_100, file_name)

            # Guardar la imagen con los valores divididos
            fits.writeto(output_path, data_divided_by_100, header=hdr, overwrite=True)

            # Cerrar el archivo FITS para liberar memoria
            hdu.close()

        except Exception as ex:
            print(f"Error processing file {file_name}: {ex}")
```
