```python
import numpy as np
import os
import pdb
from astropy.io import fits
from astropy import wcs
import subprocess
```

```python
folder_input_original_stamps = "stamps"
folder_output_stamps_with_negative_values = "stamps_negative"
folder_output_stamps_without_negative_values ="stamps_capped_negative"
```

```python
os.makedirs(folder_output_stamps_with_negative_values, exist_ok=True)
os.makedirs(folder_output_stamps_without_negative_values, exist_ok=True)
```

```python
for file_name in os.listdir(folder_input_original_stamps):
    if file_name.endswith(".fits"):
        try:
            # Open the FITS file
            file_path = os.path.join(folder_input_original_stamps, file_name)
            hdu = fits.open(file_path)
            data = hdu[0].data
            hdr = hdu[0].header

            # Calculate the median and standard deviation
            median = np.median(data)
            std_dev = np.std(data)

            # Print the median and standard deviation
            print(f"File: {file_name}, Median: {median}, Standard Deviation: {std_dev}")

            # Subtract the median from all pixels
            data_median_subtracted = data - median

            # Save the image with the median subtracted
            output_path_median_subtracted = os.path.join(folder_output_stamps_with_negative_values, f"{os.path.splitext(file_name)[0]}_median_subtracted.fits")
            fits.writeto(output_path_median_subtracted, data_median_subtracted, header=hdr, overwrite=True)

            # Replace negative pixels with 1e-10
            data_no_negatives = np.where(data_median_subtracted < 0, 1e-10, data_median_subtracted)

            # Save the image with negative pixels replaced
            output_path_no_negatives = os.path.join(folder_output_stamps_without_negative_values, f"{os.path.splitext(file_name)[0]}_no_negatives.fits")
            fits.writeto(output_path_no_negatives, data_no_negatives, header=hdr, overwrite=True)

            # Close the FITS file to free memory
            hdu.close()
            

        except Exception as ex:
            print(ex)
```
