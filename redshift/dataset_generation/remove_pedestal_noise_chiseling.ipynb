```python
import numpy as np
import os
import pdb
from astropy.io import fits
from astropy import wcs
import subprocess
```

```python
folder_input_original_stamps = "stamps"
folder_output_noise_chiseled ="stamps_noise_chiseled"
folder_output_stamps_with_negative_values = "stamps_negative"
folder_output_stamps_without_negative_values ="stamps_capped_negative"
```

```python
os.makedirs(folder_output_noise_chiseled, exist_ok=True)
os.makedirs(folder_output_stamps_with_negative_values, exist_ok=True)
os.makedirs(folder_output_stamps_without_negative_values, exist_ok=True)
```

```python
for file_name in os.listdir(folder_input_original_stamps):
    if file_name.endswith(".fits"):
        try:
            print(file_name)
            input_image_no_extension = os.path.splitext(file_name)[0]
            input_image_path = os.path.join(folder_input_original_stamps, file_name)
            command_wcslint= f"wcslint {input_image_path}"
            subprocess.run(command_wcslint, shell=True, check=True)
            noisechiseling_original = os.path.join(folder_output_noise_chiseled, input_image_no_extension + "_noisechiseling_original.fits")
            
            command = f"astnoisechisel {input_image_path} -h0 --output={noisechiseling_original}"            
            subprocess.run(command, shell=True, check=True)
        except Exception as ex:
            print(ex)
```
