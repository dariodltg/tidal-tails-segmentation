```python
import os
import glob
```

```python
path_masks_vis = "masks_train_VIS"
path_masks_NISP_H = "masks_train_NISP_H"
path_masks_NISP_J = "masks_train_NISP_J"
path_masks_NISP_Y = "masks_train_NISP_Y"


path_inputs_train_vis = "v13_log/galaxies_train_VIS"
path_inputs_test_vis = "v13_log/galaxies_test_VIS"

path_inputs_train_NISP_H = "v13_log/galaxies_train_NISP_H"
path_inputs_test_NISP_H = "v13_log/galaxies_test_NISP_H"

path_inputs_train_NISP_J = "v13_log/galaxies_train_NISP_J"
path_inputs_test_NISP_J = "v13_log/galaxies_test_NISP_J"

path_inputs_train_NISP_Y = "v13_log/galaxies_train_NISP_Y"
path_inputs_test_NISP_Y = "v13_log/galaxies_test_NISP_Y"
```

```python
def get_galaxy_number(galaxy_name:str):
    return os.path.basename(galaxy_name).split('_')[4]

def get_galaxy_magnitude(galaxy_name:str):
    return os.path.basename(galaxy_name).split('_')[5]
```

```python
i=0
j=0
for archivo in os.listdir(path_inputs_train_vis):
    numero_galaxia = get_galaxy_number(archivo)
    #print("numero_galaxia: "+ str(numero_galaxia))
    magnitud_galaxia = get_galaxy_magnitude(archivo)
    #print("magnitud_galaxia: "+ str(magnitud_galaxia))
    path_mask = os.path.join(path_masks_vis, "mask_"+str(numero_galaxia)+"_"+str(magnitud_galaxia)+".png")
    if not os.path.exists(path_mask):
        i=i+1
        print("path_mask: "+path_mask)
        #os.remove(os.path.join(path_inputs_train_vis,archivo))
    else:
        
        j = j+1
print(i)
i=0
j=0
for archivo in os.listdir(path_inputs_test_vis):
    numero_galaxia = get_galaxy_number(archivo)
    #print("numero_galaxia: "+ str(numero_galaxia))
    magnitud_galaxia = get_galaxy_magnitude(archivo)
    #print("magnitud_galaxia: "+ str(magnitud_galaxia))
    path_mask = os.path.join(path_masks_vis, "mask_"+str(numero_galaxia)+"_"+str(magnitud_galaxia)+".png")
    if not os.path.exists(path_mask):
        i=i+1
        print("path_mask: "+path_mask)
        #os.remove(os.path.join(path_inputs_test_vis,archivo))
    else:
        
        j = j+1
print(i)
```
